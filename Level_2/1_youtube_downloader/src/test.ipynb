```python
import argparse
from pathlib import Path
from pytubefix import YouTube
from pytubefix.exceptions import VideoUnavailable
from pytubefix.cli import on_progress
from pathlib import Path
```

```python
class YouTubeDownloader:
    def __init__(self, url, output_path='videos', quality=None):
        self.url = url
        self.output_path = output_path or Path.cwd()
        self.quality = quality
        self.yt = YouTube(url, on_progress_callback=on_progress)

    def download(self):
        self.yt.streams.get_highest_resolution(progressive=True).download(output_path=self.output_path)
```

```python
yt = YouTubeDownloader('https://www.youtube.com/watch?v=5qF_qbaWt3Q')
yt.download()
```
