```python
import smtplib
from email.mime.text import MIMEText
import os
```

```python
def send_simple_email(sender, recipient, subject, body):
    msg = MIMEText(body)
    msg['Subject'] = subject
    msg['From'] = sender
    msg['To'] = recipient
    # Set up the SMTP server
    smtp_server = 'smtp.gmail.com'
    smtp_port = 587
    sender_email = 'majid.zare71@gmail.com'
    # Note: you need google app password here. Use an app password instead of your Gmail password.
    # https://myaccount.google.com/apppasswords
    password = os.getenv('EMAIL_PASSWORD')  # Use an environment variable for security

    try:
        with smtplib.SMTP(smtp_server, smtp_port) as server:
            server.starttls()  # Upgrade to secure connection
            server.login(sender_email, password)  # Use environment variable for password
            server.send_message(msg)
            print("Email sent successfully!")
    except Exception as e:
        print(f"Failed to send email: {e}")

send_simple_email(
    sender='majid.zare71@gmail.com',
    recipient='majid.zare71@gmail.com',
    subject='Test Email',
    body='This is a test email sent from Python.'
)
```
