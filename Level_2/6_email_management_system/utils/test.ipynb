```python
import smtplib
from email.mime.text import MIMEText
```

```python
msg = MIMEText("This is a test email sent from Python.")
msg['Subject'] = 'Test Email'
msg['From'] = 'majid.zare71@gmail.com'
msg['To'] = 'majid.zare71@gmail.com'

server = smtplib.SMTP('smtp.gmail.com', 587)
sender_email = 'majid.zare71@gmail.com'
# Note: you need google app password here. Use an app password instead of your Gmail password.
# https://myaccount.google.com/apppasswords
email_password = 'password'  # Replace with your actual password
```

```python
server.starttls()
server.login(user=sender_email, password=email_password)
server.send_message(msg)
server.quit()
```
