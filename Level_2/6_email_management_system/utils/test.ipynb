```python
import smtplib
from email.mime.multipart import MIMEMultipart
from email.mime.text import MIMEText
from email.mime.base import MIMEBase
import os
import base64
```

```python
def send_html_email(sender, recipient, subject, text_body='', html_body='', file_path=None):
    msg = MIMEMultipart('alternative')
    msg['Subject'] = subject
    msg['From'] = sender
    msg['To'] = recipient

    # Create the plain text and HTML versions of your message
    part1 = MIMEText(text_body, 'plain')
    part2 = MIMEText(html_body, 'html')

    # Attach parts into message container
    msg.attach(part1)
    msg.attach(part2)

    # If a file path is provided, attach the file
    if file_path and os.path.isfile(file_path):
        with open(file_path, 'rb') as file:
            part3 = MIMEBase('application', 'octet-stream', Name=os.path.basename(file_path))
            part3.set_payload(file.read())
            part3.add_header('Content-Disposition', f'attachment; filename="{os.path.basename(file_path)}"')

            # Encode the payload using base64
            from email import encoders
            encoders.encode_base64(part3)
            msg.attach(part3)

    # Set up the SMTP server
    smtp_server = 'smtp.gmail.com'
    smtp_port = 587
    # Note: you need google app password here. Use an app password instead of your Gmail password.
    # https://myaccount.google.com/apppasswords
    password = os.getenv('EMAIL_PASSWORD')  # Use an environment variable for security

    try:
        with smtplib.SMTP(smtp_server, smtp_port) as server:
            server.starttls()  # Upgrade to secure connection
            server.login(sender, password)  # Use environment variable for password
            server.send_message(msg)
            print("Email sent successfully!")
    except Exception as e:
        print(f"Failed to send email: {e}")

send_html_email(
    sender='majid.zare71@gmail.com',
    recipient='majid.zare71@gmail.com',
    subject='Test Email',
    file_path="attachments/sulphate.pdf",
    text_body='''
    Hello, World!
    This is a test email with plain text content.
    Enjoy sending emails with Python!
    This is a footer text.
    ''',
    html_body='''
    <html>
    <head>
        <style>
            body { font-family: Arial, sans-serif; }
            h1 { color: #333; }
            p { font-size: 16px; }
            .container { width: 80%; margin: auto; }
            .footer { text-align: center; font-size: 12px; color: #777; }
        </style>
    </head>
    <body>
        <div class="container">
            <h1>Hello, World!</h1>
            <p>This is a test email with HTML content.</p>
            <p>Enjoy sending emails with Python!</p>
            <div class="footer">This is a footer text.</div>
        </div>
    </body>
    </html>
    '''
)
```
