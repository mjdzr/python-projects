```python
from PyPDF2 import PdfReader


def extract_from_pdf(pdf_path):
    reader = PdfReader(pdf_path)
    resume_text = ""
    for page in reader.pages:
        text = page.extract_text()
        resume_text += text

    return resume_text
```

```python
resume_text = extract_from_pdf("data/CV - Majid Zaremehrjardy.pdf")
```

```python
# Try the llm_interface module
from llm_interface import call_llm
from prompts import RESUME_JSON_SCHEMA, LLM_JSON_PARSE_PROMPT, RESUME_YAML_SCHEMA, LLM_YAML_PARSE_PROMPT, RESUME_REVIEW_PROMPT, REVIEW_OUTPUT_SCHEMA

prompt_json = LLM_JSON_PARSE_PROMPT.format(resume_text=resume_text, resume_schema=RESUME_JSON_SCHEMA)
response_json = call_llm(prompt_json)
# print(response_json)
```

```python
# Try the YAML parsing
prompt_yaml = LLM_YAML_PARSE_PROMPT.format(resume_text=resume_text, resume_schema=RESUME_YAML_SCHEMA)
response_yaml = call_llm(prompt_yaml)
```

```python
# Review
review_prompt = RESUME_REVIEW_PROMPT.format(resume_data=response_yaml, review_output_schema=REVIEW_OUTPUT_SCHEMA)
review_response = call_llm(review_prompt)
print(review_response)
```
