```python
from PyPDF2 import PdfReader


def extract_from_pdf(pdf_path):
    reader = PdfReader(pdf_path)
    resume_text = ""
    for page in reader.pages:
        text = page.extract_text()
        resume_text += text

    return resume_text
```

```python
from llm_interface import call_llm
from prompts import RESUME_YAML_SCHEMA, LLM_YAML_PARSE_PROMPT, RESUME_REVIEW_PROMPT, REVIEW_OUTPUT_SCHEMA, JOB_DESCRIPTION_REVIEW_PROMPT
from constants import EXAMPLE_JOB_DESCRIPTION

pdf_path = "data/CV - Majid Zaremehrjardy.pdf"
resume_text = extract_from_pdf(pdf_path)

# Try the YAML parsing
prompt_yaml = LLM_YAML_PARSE_PROMPT.format(resume_text=resume_text, resume_schema=RESUME_YAML_SCHEMA)
response_yaml = call_llm(prompt_yaml)
```

```python
# Review the resume
review_prompt = RESUME_REVIEW_PROMPT.format(resume_data=response_yaml, review_output_schema=REVIEW_OUTPUT_SCHEMA)
review_response = call_llm(review_prompt)
# print(review_response)
```

```python
# Review the resume based on a job description
job_description_prompt = JOB_DESCRIPTION_REVIEW_PROMPT.format(resume_data=response_yaml, 
                                                              job_description=EXAMPLE_JOB_DESCRIPTION, 
                                                              review_output_schema=REVIEW_OUTPUT_SCHEMA)
job_description_response = call_llm(job_description_prompt)
print(job_description_response)
```
