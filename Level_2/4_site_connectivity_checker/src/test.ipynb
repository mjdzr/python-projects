```python
import requests
import threading
import concurrent

def check_website_availability(url):
    try:
        header = {
            'User-Agent': 'Mozilla/5.0 (Windows NT 10.0; Win64; x64) AppleWebKit/537.36 (KHTML, like Gecko) Chrome/58.0.3029.110 Safari/537.3'
        }
        response = requests.get(url, headers=header, timeout=5)
        if response.status_code == 200:
            print(f"Website {url} is available.\n")
            return True
        else:
            print(f"Website {url} returned status code: {response.status_code}\n")
            return False
    except requests.exceptions.RequestException as e:
        print(f"Error accessing {url}: {e}\n")
        return False
```

```python
# Create a combination of websites that exists and does not exist:
websites_to_check = [
    "https://www.google.com",
    "https://www.nonexistentwebsite123456789.com",
    "https://www.github.com",
    "https://www.python.org",
    "https://www.thiswebsitedoesnotexistatall.com",
    "https://www.wikipedia.org",
    "https://www.stackoverflow.com",
    "https://www.reddit.com",
    "https://www.nonexistentwebsite.com",
    "https://www.example.com",
    "https://www.bbc.com",
    "https://www.cnn.com",
    "https://www.nytimes.com",
    "https://www.nonexistentwebsiteagain.com",
    "https://www.amazon.com"
]
```

```python
threads = []
for website in websites_to_check:
    thread = threading.Thread(target=check_website_availability, args=(website,))
    threads.append(thread)
    thread.start()
```

```python
for thread in threads:
    thread.join()
```

```python
with concurrent.futures.ThreadPoolExecutor(max_workers=5) as executor:
    results = executor.map(check_website_availability, websites_to_check)
```
